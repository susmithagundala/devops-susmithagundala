```python
import pandas as pd
import numpy as np

data = {
    "order_id": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],

    "customer_city": [
        "Chennai", " chennai ", "BENGALURU", "Hyderabad",
        "mumbai", "CHENNAI", None, "Pune", " pune ", "HYDERABAD"
    ],

    "product_name": [
        "Python Basics", "Data Science 101", "Atomic Habits",
        "Deep Work", "Rich Dad Poor Dad", "Clean Code",
        "The Alchemist", "Think and Grow Rich",
        "Ikigai", "The Psychology of Money"
    ],

    "quantity": [2, 1, 3, 1, 2, 1, 4, 2, -1, 3],

    "unit_price": [
        "₹450", "550", "₹399", "invalid", "₹299",
        "₹650", "350", "₹250", "₹499", "₹399"
    ],

    "discount_percent": [
        10, 5, np.nan, 15, 0, 10, 5, np.nan, 10, 110
    ],

    "order_date": [
        "2026-10-01", "02/10/2026", "2026/10/03",
        "04-10-2026", "2026-10-05", "06/10/2026",
        "2026-10-07", "08-10-2026", "2026/10/09",
        "10/10/2026"
    ],

    "payment_method": [
        "UPI", " credit card ", "COD", "UPI", "Debit Card",
        "upi", "Net Banking", None, "COD", "CREDIT CARD"
    ],

    "delivery_days": [3, 5, 2, 7, 4, 6, 3, 999, 5, 4],

    "order_status": [
        "Delivered", "delivered", "SHIPPED", "Delivered",
        "Cancelled", "Delivered", "Pending", "shipped",
        "CANCELLED", "Delivered"
    ]
}

df = pd.DataFrame(data)

original_df = df.copy()

df
```

```python
print(df.isnull())
```

```python
print(df.isna())
```

```python
print(df)
print("\nShape:", df.shape)
print("\nColumn data types:")
print(df.dtypes)

print("\nDetailed information:")
df.info()
```

```python
print("Missing values using isnull():")
print(df.isnull())

print("\nMissing values using isna():")
print(df.isna())

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nTotal missing cells:")
print(df.isna().sum().sum())
```

```python
df_dropped = df.dropna()

print("Original rows:", len(df))
print("Rows after dropna():", len(df_dropped))
print(df_dropped)
```

```python
df_clean = df.copy()

df_clean["customer_city"] = (
    df_clean["customer_city"].fillna("Unknown")
)

df_clean["discount_percent"] = (
    df_clean["discount_percent"].fillna(
        df_clean["discount_percent"].median()
    )
)

df_clean["payment_method"] = (
    df_clean["payment_method"].fillna("Unknown")
)

print(df_clean.isna().sum())
```

```python
print("Quantity total:", df["quantity"].sum())
print("Average quantity:", df["quantity"].mean())
print("Median quantity:", df["quantity"].median())

print("\nDelivery statistics:")
print(df["delivery_days"].describe())
```

```python
df_clean["customer_city"] = (
    df_clean["customer_city"].str.strip().str.title()
)

df_clean["product_name"] = (
    df_clean["product_name"].str.strip().str.title()
)

df_clean["payment_method"] = (
    df_clean["payment_method"].str.strip().str.title()
)

df_clean["order_status"] = (
    df_clean["order_status"].str.strip().str.title()
)

print(df_clean[[
    "customer_city",
    "product_name",
    "payment_method",
    "order_status"
]])
```

```python
df_clean["unit_price"] = pd.to_numeric(
    df_clean["unit_price"]
        .str.replace("₹", "", regex=False)
        .str.strip(),
    errors="coerce"
)

print(df_clean["unit_price"])
print(df_clean.dtypes)
```

```python
df_clean["order_date"] = pd.to_datetime(
    df_clean["order_date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print(df_clean["order_date"])
print(df_clean["order_date"].dtype)
```

```python
df_clean["unit_price"] = df_clean["unit_price"].fillna(
    df_clean["unit_price"].median()
)

print(df_clean.isna().sum())
print(df_clean.dtypes)
```

```python
print("Invalid quantities:")
print(df_clean[df_clean["quantity"] <= 0])

print("\nInvalid discounts:")
print(
    df_clean[
        (df_clean["discount_percent"] < 0) |
        (df_clean["discount_percent"] > 100)
    ]
)
```

```python
df_clean.loc[
    df_clean["quantity"] <= 0, "quantity"
] = 1

df_clean.loc[
    (df_clean["discount_percent"] < 0) |
    (df_clean["discount_percent"] > 100),
    "discount_percent"
] = np.nan

df_clean["discount_percent"] = (
    df_clean["discount_percent"].fillna(
        df_clean["discount_percent"].median()
    )
)

print(df_clean[[
    "order_id", "quantity", "discount_percent"
]])
```

```python
Q1 = df_clean["delivery_days"].quantile(0.25)
Q3 = df_clean["delivery_days"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = df_clean[
    (df_clean["delivery_days"] < lower_bound) |
    (df_clean["delivery_days"] > upper_bound)
]

print("\nPotential outliers:")
print(outliers)
```

```python
df_before_outlier = df_clean.copy()

df_clean.loc[
    df_clean["delivery_days"] == 999,
    "delivery_days"
] = 9

print(df_clean[["order_id", "delivery_days"]])
```

```python
df_clean["order_value"] = (
    df_clean["quantity"] *
    df_clean["unit_price"] *
    (1 - df_clean["discount_percent"] / 100)
).round(2)

print(df_clean[[
    "order_id",
    "product_name",
    "quantity",
    "unit_price",
    "discount_percent",
    "order_value"
]])
```

```python
print("Original DataFrame:")
print(original_df)

print("\nCleaned DataFrame:")
print(df_clean)

print("\nOriginal shape:", original_df.shape)
print("Cleaned shape:", df_clean.shape)

print("\nMissing values before cleaning:")
print(original_df.isna().sum())

print("\nMissing values after cleaning:")
print(df_clean.isna().sum())

print("\nOriginal quantity statistics:")
print(original_df["quantity"].agg(["sum", "mean", "median"]))

print("\nCleaned quantity statistics:")
print(df_clean["quantity"].agg(["sum", "mean", "median"]))

print("\nCleaned numerical summary:")
print(df_clean.describe())
```

```python
print("Total order value:", df_clean["order_value"].sum())

print("\nAverage order value:", df_clean["order_value"].mean())

print("\nRevenue by city:")
print(
    df_clean.groupby("customer_city")["order_value"]
    .sum()
    .sort_values(ascending=False)
)

print("\nOrders by payment method:")
print(df_clean["payment_method"].value_counts())

print("\nOrders by status:")
print(df_clean["order_status"].value_counts())
```

```python
df_clean.to_csv("cleaned_book_orders.csv", index=False)

print("Cleaned dataset saved successfully!")
```
