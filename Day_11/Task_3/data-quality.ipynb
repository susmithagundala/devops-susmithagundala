```python
import pandas as pd
import numpy as np

data = {
    "order_id": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],

    "customer_city": [
        "Chennai", " chennai ", "BENGALURU", "Hyderabad", "mumbai",
        "CHENNAI", None, "Pune", " pune ", "HYDERABAD"
    ],

    "product_name": [
        "Python Basics", "Data Science 101", "Atomic Habits",
        "Deep Work", "Rich Dad Poor Dad", "Clean Code",
        "The Alchemist", "Think and Grow Rich", "Ikigai",
        "The Psychology of Money"
    ],

    "quantity": [2, 1, 3, 1, 2, 1, 4, 2, -1, 3],

    "unit_price": [
        "₹450", "550", "₹399", "invalid", "₹299",
        "₹650", "350", "₹250", "₹499", "₹399"
    ],

    "discount_percent": [
        10, 5, np.nan, 15, 0, 10, 5, np.nan, 10, 110
    ],

    "order_date": [
        "2026-10-01", "02/10/2026", "2026/10/03", "04-10-2026",
        "2026-10-05", "06/10/2026", "2026-10-07", "08-10-2026",
        "2026/10/09", "10/10/2026"
    ],

    "payment_method": [
        "UPI", " credit card ", "COD", "UPI", "Debit Card",
        "upi", "Net Banking", None, "COD", "CREDIT CARD"
    ],

    "delivery_days": [3, 5, 2, 7, 4, 6, 3, 999, 5, 4],

    "order_status": [
        "Delivered", "delivered", "SHIPPED", "Delivered",
        "Cancelled", "Delivered", "Pending", "shipped",
        "CANCELLED", "Delivered"
    ]
}

df = pd.DataFrame(data)

df.head(10)
```

```python
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nSummary statistics:")
display(df.describe(include="all"))
```

```python
df.isnull()
```

```python
missing_count = df.isnull().sum()

print("Missing values in each column:")
print(missing_count)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing-value percentage:")
print(missing_percentage)
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_report = missing_report[
    missing_report["Missing_Count"] > 0
]

display(missing_report)
```

```python
df_original = df.copy()
```

```python
df_work = df.copy()

df_work["customer_city"] = (
    df_work["customer_city"].fillna("Unknown")
)

df_work["payment_method"] = (
    df_work["payment_method"].fillna("Unknown")
)

df_work["discount_percent"] = (
    df_work["discount_percent"].fillna(
        df_work["discount_percent"].median()
    )
)

print("Missing values after handling:")
print(df_work.isnull().sum())
```

```python
# Check every row for duplicates
duplicate_check = df.duplicated()

print(duplicate_check)

# Count duplicate rows
print("\nTotal duplicate rows:", df.duplicated().sum())
```

```python
# Find repeated order IDs
repeated_ids = df["order_id"].duplicated(keep=False)

print("Records with repeated order IDs:")
display(df[repeated_ids])

print("Number of records with repeated order IDs:", repeated_ids.sum())
```

```python
order_counts = df["order_id"].value_counts()

print("Order ID occurrences:")
print(order_counts)

print("\nOrder IDs appearing more than once:")
print(order_counts[order_counts > 1])
```

```python
duplicate_records = df[
    df.duplicated(subset="order_id", keep=False)
].sort_values("order_id")

display(duplicate_records)
```

```python
# Create a separate DataFrame for analysis
df_analysis = df.copy()

# Remove the rupee symbol and extra spaces
df_analysis["unit_price"] = (
    df_analysis["unit_price"]
    .str.replace("₹", "", regex=False)
    .str.strip()
)

# Convert prices to numeric values
df_analysis["unit_price"] = pd.to_numeric(
    df_analysis["unit_price"],
    errors="coerce"
)

print(df_analysis[["product_name", "unit_price"]])
```

```python
# Select records suitable for revenue analysis
valid_orders = df_analysis[
    df_analysis["unit_price"].notna()
    & (df_analysis["quantity"] > 0)
    & df_analysis["discount_percent"].between(0, 100)
].copy()

# Calculate order value
valid_orders["order_value"] = (
    valid_orders["quantity"]
    * valid_orders["unit_price"]
    * (1 - valid_orders["discount_percent"] / 100)
).round(2)

display(
    valid_orders[
        ["order_id", "customer_city", "product_name",
         "quantity", "unit_price", "discount_percent", "order_value"]
    ]
)
```

```python
city_analysis = (
    valid_orders.groupby("customer_city")
    .agg(
        total_orders=("order_id", "count"),
        total_quantity=("quantity", "sum"),
        total_revenue=("order_value", "sum"),
        average_order_value=("order_value", "mean"),
        median_order_value=("order_value", "median")
    )
    .round(2)
    .sort_values("total_revenue", ascending=False)
)

display(city_analysis)
```

```python
print("Highest revenue:")
display(city_analysis.head(1))

print("Lowest revenue:")
display(city_analysis.tail(1))
```

```python
highest_price = df_analysis["unit_price"].max()

print("Highest unit price: ₹", highest_price)
```

```python
highest_index = df_analysis["unit_price"].idxmax()

highest_priced_item = df_analysis.loc[highest_index]

print("Highest-priced item details:")
print(highest_priced_item)
```

```python
top_five = df_analysis.nlargest(
    5, "unit_price"
)[["product_name", "unit_price", "quantity", "customer_city"]]

print("Top 5 highest-priced books:")
display(top_five)
```

```python
print("Highest price:", df_analysis["unit_price"].max())

print("\nPrice summary:")
print(df_analysis["unit_price"].describe())

print("\nRecords with invalid or missing prices:")
display(df_analysis[df_analysis["unit_price"].isna()])
```
